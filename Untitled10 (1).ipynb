```python

!pip -q install -U ultralytics opencv-python-headless pyyaml

import os
import random
import shutil
import zipfile
from pathlib import Path

import cv2
import numpy as np
import yaml
import torch
from PIL import Image, ImageDraw
from IPython.display import display
from google.colab import files
from ultralytics import YOLO

# Settings
SEED = 42
IMG_SIZE = 640
EPOCHS = 120
PATIENCE = 30
MODEL_NAME = "yolo26s.pt"  # pretrained small model
GRAY_PROB = 0.20            # grayscale copies for some training images
WATER_PROB = 0.60           # water-effect copies for some training images
BATCH_GPU = 16
BATCH_CPU = 4

random.seed(SEED)
np.random.seed(SEED)
rng = np.random.default_rng(SEED)

# Load the dataset
# Use an extracted dataset if one is already here; otherwise ask for the ZIP.
search_dirs = [Path("/content/mate_crabs_yolo_dataset"), Path("/content/dataset")]
yaml_path = None
for base in search_dirs:
    if base.exists():
        candidates = list(base.rglob("dataset.yaml")) + list(base.rglob("data.yaml"))
        if candidates:
            yaml_path = candidates[0]
            break

if yaml_path is None:
    print("Upload the generated DATASET ZIP (for example: mate_crabs_yolo_dataset.zip).")
    uploaded = files.upload()
    zip_names = [name for name in uploaded if name.lower().endswith(".zip")]
    if not zip_names:
        raise RuntimeError("No ZIP file uploaded. Upload the generated dataset ZIP, not the code bundle.")

    zip_name = zip_names[0]
    zip_path = Path("/content") / zip_name
    extract_dir = Path("/content/mate_crabs_dataset_extract")
    if extract_dir.exists():
        shutil.rmtree(extract_dir)
    extract_dir.mkdir(parents=True, exist_ok=True)

    with zipfile.ZipFile(zip_path, "r") as archive:
        archive.extractall(extract_dir)

    candidates = list(extract_dir.rglob("dataset.yaml")) + list(extract_dir.rglob("data.yaml"))
    if not candidates:
        raise RuntimeError(
            "I could not find dataset.yaml/data.yaml inside this ZIP. "
            "You may have uploaded the generator-code ZIP instead of the generated dataset ZIP."
        )
    yaml_path = candidates[0]

dataset_root = yaml_path.parent.resolve()
print("Dataset folder:", dataset_root)

with open(yaml_path, "r", encoding="utf-8") as f:
    original_cfg = yaml.safe_load(f) or {}

names_value = original_cfg.get("names")
if isinstance(names_value, dict):
    class_names = [names_value[k] for k in sorted(names_value, key=lambda x: int(x))]
elif isinstance(names_value, list):
    class_names = names_value
else:
    raise RuntimeError("dataset.yaml must contain 'names' as a list or dictionary.")

if not class_names:
    raise RuntimeError("No classes were found in dataset.yaml.")

# Check the labels
IMAGE_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

def image_files(folder):
    return sorted(p for p in folder.iterdir() if p.is_file() and p.suffix.lower() in IMAGE_EXTS)

def corresponding_label(image_path, split):
    return dataset_root / "labels" / split / (image_path.stem + ".txt")

def check_split(split):
    img_dir = dataset_root / "images" / split
    label_dir = dataset_root / "labels" / split
    if not img_dir.exists() or not label_dir.exists():
        raise RuntimeError(f"Missing images/{split} or labels/{split} folder.")

    imgs = image_files(img_dir)
    if not imgs:
        raise RuntimeError(f"No images found in images/{split}.")

    missing = []
    invalid = []
    box_count = 0
    for im in imgs:
        label_path = corresponding_label(im, split)
        if not label_path.exists():
            missing.append(im.name)
            continue

        text = label_path.read_text(encoding="utf-8").strip()
        if not text:
            continue  # empty label file is valid for a negative image

        for line_no, line in enumerate(text.splitlines(), start=1):
            parts = line.split()
            try:
                if len(parts) != 5:
                    raise ValueError("expected 5 fields")
                cls = int(parts[0])
                cx, cy, w, h = map(float, parts[1:])
                good = (
                    0 <= cls < len(class_names) and
                    all(np.isfinite(v) for v in [cx, cy, w, h]) and
                    0 <= cx <= 1 and 0 <= cy <= 1 and
                    0 < w <= 1 and 0 < h <= 1 and
                    cx - w / 2 >= -1e-4 and cx + w / 2 <= 1.0001 and
                    cy - h / 2 >= -1e-4 and cy + h / 2 <= 1.0001
                )
                if not good:
                    raise ValueError("class id or normalized box is out of range")
                box_count += 1
            except Exception as e:
                invalid.append(f"{label_path.name}:{line_no} ({e})")

    if missing:
        raise RuntimeError(
            f"{split}: {len(missing)} images have no matching .txt label files. "
            f"Examples: {missing[:5]}. Fix labels before training."
        )
    if invalid:
        raise RuntimeError(
            f"{split}: invalid YOLO label rows were found. Examples: {invalid[:8]}"
        )

    print(f"{split:>5}: {len(imgs)} images | {box_count} boxes | numeric labels OK")
    return imgs

print("\nChecking all original labels (this checks coordinates/format, not whether boxes visually fit the crabs):")
train_images = check_split("train")
val_images = check_split("val")
test_images = check_split("test")

# Draw boxes on a few training images so they can be checked by eye.
# Valid coordinates do not guarantee that a box fits the crab itself.
preview_candidates = train_images[:6]
preview_sheet = Image.new("RGB", (960, 700), "white")
for i, image_path in enumerate(preview_candidates):
    im = Image.open(image_path).convert("RGB")
    draw = ImageDraw.Draw(im)
    label_path = corresponding_label(image_path, "train")
    label_text = label_path.read_text(encoding="utf-8").strip()
    for line in label_text.splitlines() if label_text else []:
        parts = line.split()
        cls = int(parts[0])
        cx, cy, bw, bh = map(float, parts[1:])
        x1 = int((cx - bw / 2) * im.width)
        y1 = int((cy - bh / 2) * im.height)
        x2 = int((cx + bw / 2) * im.width)
        y2 = int((cy + bh / 2) * im.height)
        draw.rectangle((x1, y1, x2, y2), outline=(0, 255, 0), width=3)
        draw.text((x1, max(0, y1 - 14)), class_names[cls], fill=(0, 255, 0))
    im.thumbnail((310, 310))
    x = (i % 3) * 320
    y = (i // 3) * 345
    preview_sheet.paste(im, (x + 5, y + 20))
    ImageDraw.Draw(preview_sheet).text((x + 5, y + 3), image_path.name, fill="black")
print("\nIMPORTANT: inspect the green boxes below. They must tightly fit the actual target crabs, not surrounding paper/background. If not, stop and fix/rebuild labels before training.")
display(preview_sheet)

# Make grayscale and water-effect copies
# These effects change color and image quality, not the box positions.
# Only make extra copies for training. Leave validation and test alone.
train_aug_img = dataset_root / "images" / "train_aug"
train_aug_lbl = dataset_root / "labels" / "train_aug"
for p in (train_aug_img, train_aug_lbl):
    if p.exists():
        shutil.rmtree(p)
    p.mkdir(parents=True, exist_ok=True)

def water_filter_bgr(image_bgr, local_rng):
    """Give a BGR image a mild underwater look."""
    x = image_bgr.astype(np.float32)

    # OpenCV uses BGR order, so the red channel is index 2.
    x[:, :, 2] *= float(local_rng.uniform(0.62, 0.88))
    x[:, :, 1] *= float(local_rng.uniform(0.90, 1.04))
    x[:, :, 0] *= float(local_rng.uniform(1.00, 1.15))

    # Add a little blue-green haze without hiding too much detail.
    haze = float(local_rng.uniform(0.03, 0.13))
    haze_bgr = np.array([155, 148, 112], dtype=np.float32)
    x = x * (1.0 - haze) + haze_bgr[None, None, :] * haze

    # Vary the brightness.
    x *= float(local_rng.uniform(0.78, 1.12))
    x = np.clip(x, 0, 255).astype(np.uint8)

    if local_rng.random() < 0.40:
        x = cv2.GaussianBlur(x, (0, 0), float(local_rng.uniform(0.25, 0.85)))

    noise_std = float(local_rng.uniform(0.3, 2.2))
    noise = local_rng.normal(0, noise_std, x.shape).astype(np.float32)
    return np.clip(x.astype(np.float32) + noise, 0, 255).astype(np.uint8)

n_gray = 0
n_water = 0
n_train_total = 0

for i, img_path in enumerate(train_images):
    label_path = corresponding_label(img_path, "train")
    # Keep the original image and its label.
    shutil.copy2(img_path, train_aug_img / img_path.name)
    shutil.copy2(label_path, train_aug_lbl / label_path.name)
    n_train_total += 1

    # Use a repeatable random sequence for each image.
    local_rng = np.random.default_rng(SEED + i * 7919 + 17)
    image = cv2.imread(str(img_path), cv2.IMREAD_COLOR)
    if image is None:
        raise RuntimeError(f"Could not read image: {img_path}")

    if local_rng.random() < GRAY_PROB:
        gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
        gray3 = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)
        out_name = img_path.stem + "__gray.jpg"
        if not cv2.imwrite(str(train_aug_img / out_name), gray3):
            raise RuntimeError(f"Could not write grayscale variant: {out_name}")
        shutil.copy2(label_path, train_aug_lbl / (Path(out_name).stem + ".txt"))
        n_gray += 1
        n_train_total += 1

    if local_rng.random() < WATER_PROB:
        water = water_filter_bgr(image, local_rng)
        out_name = img_path.stem + "__water.jpg"
        if not cv2.imwrite(str(train_aug_img / out_name), water):
            raise RuntimeError(f"Could not write water variant: {out_name}")
        shutil.copy2(label_path, train_aug_lbl / (Path(out_name).stem + ".txt"))
        n_water += 1
        n_train_total += 1

print("\nTraining-only photometric variants created:")
print(f"  Original train images: {len(train_images)}")
print(f"  Grayscale variants:    {n_gray}")
print(f"  Water-filter variants: {n_water}")
print(f"  Total train images:    {n_train_total}")
print("Validation and test were not augmented or copied into train.")

# Show a few examples of the image effects.
sample_img = train_images[0]
original_bgr = cv2.imread(str(sample_img), cv2.IMREAD_COLOR)
local_rng = np.random.default_rng(SEED + 2026)
gray_preview = cv2.cvtColor(cv2.cvtColor(original_bgr, cv2.COLOR_BGR2GRAY), cv2.COLOR_GRAY2BGR)
water_preview = water_filter_bgr(original_bgr, local_rng)

preview = Image.new("RGB", (960, 350), "white")
for j, (title, arr) in enumerate([
    ("Original", original_bgr),
    ("Grayscale", gray_preview),
    ("Water effect", water_preview),
]):
    im = Image.fromarray(cv2.cvtColor(arr, cv2.COLOR_BGR2RGB))
    im.thumbnail((310, 310))
    x = j * 320
    preview.paste(im, (x + 5, 25))
    ImageDraw.Draw(preview).text((x + 5, 5), title, fill="black")
display(preview)

# Set up the training YAML
training_yaml = dataset_root / "dataset_training_aug.yaml"
train_cfg = {
    "path": str(dataset_root),
    "train": "images/train_aug",
    "val": "images/val",
    "test": "images/test",
    "nc": len(class_names),
    "names": class_names,
}
with open(training_yaml, "w", encoding="utf-8") as f:
    yaml.safe_dump(train_cfg, f, sort_keys=False, allow_unicode=True)

# Train YOLO
has_gpu = torch.cuda.is_available()
device = 0 if has_gpu else "cpu"
batch = BATCH_GPU if has_gpu else BATCH_CPU

print("\nCompute device:", torch.cuda.get_device_name(0) if has_gpu else "CPU (training will be slow)")
if not has_gpu:
    print("Recommended: Colab Runtime > Change runtime type > T4 GPU, then rerun this cell.")

model = YOLO(MODEL_NAME)  # Start from pretrained weights.
run_project = "/content/runs/detect"
run_name = "mate_crabs_yolo26s_aug"

train_results = model.train(
    data=str(training_yaml),
    epochs=EPOCHS,
    imgsz=IMG_SIZE,
    batch=batch,
    device=device,
    workers=2,
    pretrained=True,
    optimizer="AdamW",
    lr0=0.001,
    lrf=0.01,
    weight_decay=0.0005,
    warmup_epochs=3.0,
    cos_lr=True,
    patience=PATIENCE,
    degrees=180.0,
    translate=0.08,
    scale=0.30,
    shear=1.0,
    perspective=0.0003,
    fliplr=0.5,
    flipud=0.25,
    hsv_h=0.015,
    hsv_s=0.55,
    hsv_v=0.35,
    mosaic=0.35,
    mixup=0.0,
    close_mosaic=10,
    amp=has_gpu,
    cache=False,
    plots=True,
    save=True,
    save_period=10,
    project=run_project,
    name=run_name,
    exist_ok=True,
    verbose=True,
)

run_dir = Path(run_project) / run_name
best_weights = run_dir / "weights" / "best.pt"
if not best_weights.exists():
    raise RuntimeError(f"Training finished but best.pt was not found at {best_weights}")

# Check the best model on the test split
print("\nEvaluating on the dataset TEST split...")
print("Note: if test images were generated from the same three source photos, this is not a substitute for testing on real unseen board footage.")
best_model = YOLO(str(best_weights))
test_metrics = best_model.val(
    data=str(training_yaml),
    split="test",
    imgsz=IMG_SIZE,
    batch=batch,
    device=device,
    plots=True,
    project=run_project,
    name=run_name + "_test",
    exist_ok=True,
)

try:
    print(f"Test mAP50:     {float(test_metrics.box.map50):.4f}")
    print(f"Test mAP50-95:  {float(test_metrics.box.map):.4f}")
    print(f"Test Precision: {float(test_metrics.box.mp):.4f}")
    print(f"Test Recall:    {float(test_metrics.box.mr):.4f}")
except Exception:
    print("Test metrics were produced; see Ultralytics output files for full details.")

# Save the model and training reports
bundle_dir = Path("/content/mate_crab_training_output")
if bundle_dir.exists():
    shutil.rmtree(bundle_dir)
bundle_dir.mkdir(parents=True, exist_ok=True)

shutil.copy2(best_weights, bundle_dir / "best.pt")
shutil.copy2(training_yaml, bundle_dir / "dataset_training_aug.yaml")

for filename in [
    "results.csv", "results.png", "args.yaml", "confusion_matrix.png",
    "confusion_matrix_normalized.png", "PR_curve.png", "F1_curve.png",
    "P_curve.png", "R_curve.png", "labels.jpg", "labels_correlogram.jpg"
]:
    source = run_dir / filename
    if source.exists():
        shutil.copy2(source, bundle_dir / filename)

test_run_dir = Path(run_project) / (run_name + "_test")
for filename in ["confusion_matrix.png", "confusion_matrix_normalized.png", "PR_curve.png", "F1_curve.png", "P_curve.png", "R_curve.png"]:
    source = test_run_dir / filename
    if source.exists():
        shutil.copy2(source, bundle_dir / ("test_" + filename))

(bundle_dir / "README.txt").write_text(
    "MATE crab YOLO training output\n"
    "best.pt is the trained model checkpoint.\n"
    "Review the metrics/plots and visually test detections on real board images/video.\n"
    "Synthetic test metrics may overestimate real competition performance.\n",
    encoding="utf-8"
)

output_zip = Path("/content/mate_crab_training_output.zip")
if output_zip.exists():
    output_zip.unlink()
with zipfile.ZipFile(output_zip, "w", zipfile.ZIP_DEFLATED, compresslevel=3) as z:
    for p in bundle_dir.rglob("*"):
        if p.is_file():
            z.write(p, arcname=str(Path(bundle_dir.name) / p.relative_to(bundle_dir)))

print("\nDONE. Downloading best weights + training reports:", output_zip)
files.download(str(output_zip))
```
