```python
import io
import zipfile
from pathlib import Path

import cv2
import torch
from PIL import Image
from IPython.display import display
from google.colab import files
from ultralytics import YOLO

# Settings
IMG_SIZE = 640
CONFIDENCE = 0.25
IOU_THRESHOLD = 0.50

VIDEO_EXTS = {
    ".mp4", ".avi", ".mov", ".mkv",
    ".webm", ".mpeg", ".mpg"
}

# 1. Upload the model

print("Step 1/2: Upload best.pt or mate_crab_training_output.zip")

model_upload = files.upload()

model_path = None

for filename, content in model_upload.items():
    suffix = Path(filename).suffix.lower()

    if suffix == ".pt":
        model_path = "/content/mate_uploaded_best.pt"

        with open(model_path, "wb") as f:
            f.write(content)

        break

    elif suffix == ".zip":
        try:
            with zipfile.ZipFile(io.BytesIO(content)) as archive:
                candidates = [
                    name for name in archive.namelist()
                    if Path(name).name == "best.pt"
                ]

                if not candidates:
                    candidates = [
                        name for name in archive.namelist()
                        if Path(name).suffix.lower() == ".pt"
                    ]

                if candidates:
                    model_path = "/content/mate_uploaded_best.pt"

                    with open(model_path, "wb") as f:
                        f.write(archive.read(candidates[0]))

                    break

        except zipfile.BadZipFile:
            raise RuntimeError(
                "The uploaded ZIP is not valid. "
                "Upload the training output ZIP or best.pt."
            )

if model_path is None:
    raise RuntimeError(
        "No model was found. Upload best.pt or the training output ZIP."
    )

print("\nLoading the trained model...")

model = YOLO(model_path)

# Find the Green Crab class
raw_names = model.names

if isinstance(raw_names, dict):
    class_names = {
        int(k): str(v) for k, v in raw_names.items()
    }
else:
    class_names = {
        i: str(v) for i, v in enumerate(raw_names)
    }

green_ids = [
    class_id for class_id, name in class_names.items()
    if "green" in name.lower() and "crab" in name.lower()
]

if green_ids:
    target_id = green_ids[0]
elif len(class_names) == 1:
    target_id = next(iter(class_names))
else:
    raise RuntimeError(
        f"Could not identify the Green Crab class: {class_names}"
    )

print("Model loaded successfully.")
print("Target class:", class_names[target_id])

device = 0 if torch.cuda.is_available() else "cpu"
print("Device:", "GPU" if torch.cuda.is_available() else "CPU")



print("\nStep 2/2: Now upload your ROV video only.")

video_upload = files.upload()

video_names = [
    name for name in video_upload
    if Path(name).suffix.lower() in VIDEO_EXTS
]

if not video_names:
    raise RuntimeError(
        "No supported video found. Upload an MP4, AVI, MOV, or MKV video."
    )

video_name = video_names[0]
video_path = "/content/mate_rov_input" + Path(video_name).suffix.lower()

with open(video_path, "wb") as f:
    f.write(video_upload[video_name])

print("Video loaded:", video_name)



cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise RuntimeError(
        "Could not open the video. Try exporting it as MP4."
    )

fps = cap.get(cv2.CAP_PROP_FPS)

if not fps or fps <= 0:
    fps = 30.0

total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

best_frame = None
best_count = -1
best_confidence = -1.0
best_frame_number = 0
best_time = 0.0

top_frames = []
frame_number = 0

print("\nAnalyzing the video frame by frame...")

# 4. Detect and count Green Crabs

while True:
    ok, frame = cap.read()

    if not ok:
        break

    result = model.predict(
        source=frame,
        imgsz=IMG_SIZE,
        conf=CONFIDENCE,
        iou=IOU_THRESHOLD,
        classes=[target_id],
        device=device,
        verbose=False
    )[0]

    boxes = result.boxes

    if boxes is not None and len(boxes) > 0:
        count = len(boxes)
        confidence_sum = float(boxes.conf.sum().item())
        mean_confidence = confidence_sum / count
    else:
        count = 0
        confidence_sum = 0.0
        mean_confidence = 0.0

    # Draw boxes and class names
    annotated = result.plot()

    timestamp = frame_number / fps

    cv2.putText(
        annotated,
        f"European Green Crab: {count}",
        (20, 42),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.9,
        (0, 255, 0),
        2,
        cv2.LINE_AA
    )

    candidate = {
        "frame": frame_number,
        "time": timestamp,
        "count": count,
        "confidence": confidence_sum,
        "mean_confidence": mean_confidence
    }

    top_frames.append(candidate)

    top_frames.sort(
        key=lambda x: (x["count"], x["confidence"]),
        reverse=True
    )

    top_frames = top_frames[:10]

    # Keep the frame with the highest count.
    # If two frames have the same count, prefer higher confidence.
    if (
        count > best_count
        or (
            count == best_count
            and confidence_sum > best_confidence
        )
    ):
        best_count = count
        best_confidence = confidence_sum
        best_frame_number = frame_number
        best_time = timestamp
        best_frame = annotated.copy()

    frame_number += 1

    if frame_number % 100 == 0:
        print(
            f"Processed {frame_number}/{total_frames} frames | "
            f"Highest count so far: {best_count}"
        )

cap.release()

if frame_number == 0 or best_frame is None:
    raise RuntimeError("No frames were processed.")

# 5. Display the best candidate frame

print("\n========== BEST CANDIDATE FRAME ==========")
print("Frame number:", best_frame_number)
print(f"Time: {best_time:.2f} seconds")
print("Detected Green Crabs:", best_count)

best_image = Image.fromarray(
    cv2.cvtColor(best_frame, cv2.COLOR_BGR2RGB)
)

best_image.thumbnail((1200, 900))

display(best_image)

# 6. Display the top candidate frames

print("\n========== TOP 10 CANDIDATE FRAMES ==========")

print(
    f"{'Frame':>8} {'Time (s)':>10} "
    f"{'Count':>8} {'Mean confidence':>17}"
)

for row in top_frames:
    print(
        f"{row['frame']:>8} "
        f"{row['time']:>10.2f} "
        f"{row['count']:>8} "
        f"{row['mean_confidence']:>17.3f}"
    )

print("\nVideo analysis completed.")
print("All results are displayed in this cell.")
```
